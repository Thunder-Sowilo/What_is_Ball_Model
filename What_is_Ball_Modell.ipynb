```python
from ultralytics import YOLO

# Load a model
model = YOLO("yolo11n.pt")

# Train the model
model.train(
    data="/content/drive/MyDrive/asd/Mid/data.yaml",  # path to dataset YAML
    epochs=20,  # number of training epochs
    imgsz=640,  # image size for training
    batch=16,  # batch size
    lr0=0.001,  # initial learning rate
    device="0"  # device to train on 0 (GPU) หรือ cpu
)


# ประเมินโมเดล (Evaluate model performance on the validation set)
metrics = model.val()


# บันทึกโมเดลที่ฝึกเสร็จแล้ว
model.save('best.pt')
```

```python
# ทดสอบการตรวจจับภาพจากวัตถุ ด้วย Model ที่ Train เสร็จแล้ว
from ultralytics import YOLO
import glob
import os
from IPython.display import Image, display
from PIL import Image as PILImage

# โหลดโมเดลที่ดีที่สุดจากการฝึกสอน (best.pt)
model = YOLO('/content/best.pt')

# ระบุโฟลเดอร์ย่อย 'images' ที่เก็บรูปภาพสำหรับทดสอบ
source_folder = '/content/drive/MyDrive/asd/test/'

# 1. ตรวจสอบความถูกต้องของโฟลเดอร์ก่อนรันโมเดล
if not os.path.exists(source_folder):
    print(f"❌ ไม่พบโฟลเดอร์: {source_folder}")
    print("กรุณาตรวจสอบว่าพาธใน Google Drive ถูกต้องหรือไม่ หรือลองเชื่อมต่อ Drive ใหม่อีกครั้ง")
    # แนะนำโฟลเดอร์ที่มีอยู่ใกล้เคียงให้ผู้ใช้ทราบ
    parent_dir = os.path.dirname(source_folder)
    if os.path.exists(parent_dir):
        print(f"📁 โฟลเดอร์ที่มีอยู่ในไดเรกทอรีด้านบน ({parent_dir}): {os.listdir(parent_dir)}")
else:
    # ค้นหารูปภาพภายในโฟลเดอร์
    supported_formats = ('.png', '.jpg', '.jpeg', '.webp', '.bmp')
    test_images = [f for f in os.listdir(source_folder) if f.lower().endswith(supported_formats)]

    if len(test_images) == 0:
        print(f"❌ พบโฟลเดอร์ {source_folder} แต่ไม่พบไฟล์รูปภาพที่รองรับเลย")
        print(f"ไฟล์ทั้งหมดที่ตรวจพบในโฟลเดอร์: {os.listdir(source_folder)[:10]}")
    else:
        print(f"✅ พบไฟล์รูปภาพทดสอบจำนวน {len(test_images)} รูป เริ่มการทำนายผลด้วย Model best.pt...")

        # 2. ให้โมเดลทำการทำนาย (Predict) รูปภาพทั้งหมดในโฟลเดอร์
        results = model.predict(source=source_folder, save=True, conf=0.25)

        print("\n===== การทดสอบเสร็จสมบูรณ์ =====")
        if len(results) > 0:
            save_dir = results[0].save_dir
            print(f"คุณสามารถเข้าไปดูรูปภาพผลลัพธ์ทั้งหมดได้ที่โฟลเดอร์: {save_dir}\n")
            print("===== แสดงผลลัพธ์รูปภาพ =====")

            # ดึงรายชื่อไฟล์รูปภาพจากโฟลเดอร์ที่บันทึกผลลัพธ์
            image_paths = glob.glob(f"{save_dir}/*.*")

            pdf_images = []

            # 3. วนลูปแสดงผลรูปภาพทั้งหมด
            for img_path in image_paths:
                if img_path.lower().endswith(('.png', '.jpg', '.jpeg')):
                    print(f"\nรูปภาพ: {os.path.basename(img_path)}")
                    display(Image(filename=img_path))

                    # เตรียมรูปภาพสำหรับบันทึกเป็น PDF
                    img = PILImage.open(img_path).convert('RGB')
                    pdf_images.append(img)

            # 4. บันทึกรูปภาพทั้งหมดลงในไฟล์ PDF ไฟล์เดียว
            if pdf_images:
                pdf_path = "/content/prediction_results.pdf"
                pdf_images[0].save(pdf_path, save_all=True, append_images=pdf_images[1:])
                print(f"\n===== สร้างไฟล์ PDF สำเร็จ =====")
                print(f"ไฟล์ PDF ถูกบันทึกไว้ที่: {pdf_path}")
```
